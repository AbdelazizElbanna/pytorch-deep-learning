# One Of The Most Common Errors in Deep Learning (Shape Errors)

````python
# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A, tensor_B) # (this will error)
````

We can make matrix multiplication work between tensor_A and tensor_B by making their inner dimensions match.

One of the ways to do this is with a transpose (switch the dimensions of a given tensor).

You can perform transposes in PyTorch using either:
- torch.transpose(input, dim0, dim1) - where input is the desired tensor to transpose and dim0 and dim1 are the dimensions to be swapped.
- tensor.T - where tensor is the desired tensor to transpose.

````python
# View tensor_A and tensor_B.T
print(tensor_A)
print(tensor_B.T)
print(torch.transpose(tensor_B, dim0=0, dim1=1))
````

````python
# The operation works when tensor_B is transposed
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}\n")
print(f"New shapes: tensor_A = {tensor_A.shape} (same as above), tensor_B.T = {tensor_B.T.shape}\n")
print(f"Multiplying: {tensor_A.shape} * {tensor_B.T.shape} <- inner dimensions match\n")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output) 
print(f"\nOutput shape: {output.shape}")
````

````python
# torch.mm is a shortcut for matmul
torch.mm(tensor_A, tensor_B.T)
````

````python
# playing around Linear Layers
# Since the linear layer starts with a random weights matrix, let's make it reproducible (more on this later)
torch.manual_seed(42)

# This uses matrix multiplication
linear = torch.nn.Linear(in_features=2,          # in_features = matches inner dimension of input 
                        out_features=6          # out_features = describes outer value
                        )

x = tensor_A
output = linear(x)

print(f"shape of x: {x.shape}\n")
print(f"Output: {output}, \n\nOutput shape: {output.shape}")
````

````python
linear = torch.nn.Linear(in_features=3,
                        out_features= 6)

x = tensor_A
output = linear(x)

print(f"output: {output}, \n\nOutput shape: {output.shape}")
````

````python
linear = torch.nn.Linear(in_features=3,
                        out_features= 6)

x = torch.transpose(tensor_A, dim0=0, dim1=1)       # or `tensor_A.T`
output = linear(x)

print(f"output: {output}, \n\nOutput shape: {output.shape}")
````

## Reshaping, stacking, squeezing and unsqueezing

| Method | One-line description |
| :--- | :--- |
| `torch.reshape(input, shape)` | Reshapes `input` to `shape` (if compatible), can also use `torch.Tensor.reshape()`. |
| `Tensor.view(shape)` | Returns a view of the original tensor in a different `shape` but shares the same data as the original tensor. |
| `torch.stack(tensors, dim=0)` | Concatenates a sequence of `tensors` along a new dimension ( `dim` ), all `tensors` must be same size. |
| `torch.squeeze(input)` | Squeezes `input` to remove all the dimensions with value `1`. |
| `torch.unsqueeze(input, dim)` | Returns `input` with a dimension value of `1` added at `dim`. |
| `torch.permute(input, dims)` | Returns a *view* of the original `input` with its dimensions permuted (rearranged) to `dims`. |

````python
x = torch.arange(1., 8.)

x, x.shape                  # Row tensor has one dimension 
````

````python
# Reshape the tensor
x_reshaped = x.reshape(1, 7)

x_reshaped, x_reshaped.shape  # Reshaped column tensor has 2d
````

````python
# Change the `View` (keeps same data as the original but changes the view)
# Can see more details in : https://stackoverflow.com/a/54507446/7900723        # Difference between view() and reshape(), and their way for saving in memory


z = x.view(1, 7)

z, z.shape
````

***Note: Changing any value in the tensor changes the same value in the view() tensor, mean while may or may not be changed with the reshaped one based on if reshape took a copy or not cuz it's dynamic for taking a copy***

````python
# Changing z changes x
z[:, 0] = 5
z, x
````

````python
# Stacking vertically 
x_v_stacked = torch.stack([x, x, x, x, x], dim=0)\

x_v_stacked, x.shape, x_v_stacked.shape
````

````python
# Stacking H 
x_h_stacked = torch.stack([x, x, x, x, x], dim=1)

x_h_stacked, x_h_stacked.shape
````

````python
# Squeeze: `torch.squeeze()` (remember this as squeezing the tensor to only have dimensions over 1).
print(f"previous tensor {x_reshaped}")
print(f"previous tensor shape: {x_reshaped.shape}\n")

# Remove extra dimension from x_reshaped
x_squeezed = x_reshaped.squeeze()

print(f"New tensor: {x_squeezed}")
print(f"New tensor shape: {x_squeezed.shape}")
````

````python
# Unsqueeze
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
````

````python
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
````

````python
# permute(): rearrange the order of the axes values where the input gets turned into a `view` with the new dims

x_original = torch.rand(size=(224, 12, 3))

# permute the original tensor to rearrange the axes order
x_permuted = torch.permute(x_original, dims=(2, 0, 1))      # axis0 = 2, axis1=0, axis3=1

print(f"The original shape: {x_original.shape}")
print(f"The permuted shape: {x_permuted.shape}")            # Returns a view of the data, so changing in the original means in the permuted and vice versa
````

# Pytorch tensors & Numpy

`torch.from_numpy(ndarray)` - NumPy array $\rightarrow$ PyTorch tensor.
<br>`torch.Tensor.numpy()` - PyTorch tensor $\rightarrow$ NumPy array.

````python
# From Numpy--> ndarray to torch--> tensor

array = np.arange(1., 8.)
tensor = torch.from_numpy(array)

print(f"array: {array},\n it's type: {type(array)}")
print(f"\n\n {tensor}, \n ti's type: {type(tensor)}")
````

> **Note**: By default, NumPy arrays are created with the datatype float64 and if you convert it to a PyTorch tensor, it'll keep the same datatype (as above).<br>
> However, many PyTorch calculations default to using float32.<br>
> So if you want to convert your NumPy array (float64) -> PyTorch tensor (float64) -> PyTorch tensor (float32), you can use tensor = torch.from_numpy(array).type(torch.float32).<br>

````python
# From torch--> tensor to Numpy--> ndarray

tensor = torch.ones(7)          # creates a tensor of ones with default dtype <float32>
numpy_tensor = tensor.numpy()   # will be ndarray of <float32> unless changing

tensor, tensor.dtype, numpy_tensor, numpy_tensor.dtype
````

**Note**: Changing the new `tensor` converted from Numpy `ndarray` or the Numpy `ndarray` convert from the `tensor` the other one still the same, cuz you already created a new one

# Reproducibility (Trying to take random out of random)

you create an algorithm capable of achieving X performance.<br>
And then your friend tries it out to verify you're not crazy.<br>
How could they do such a thing?

That's where reproducibility comes in.

In other words, can you get the same (or very similar) results on your computer running the same code as I get on mine?

````python
# Create two Random tensors (they suppose to be different)
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(f"Tensor A:\n{random_tensor_A}")
print(f"\nTensor B:\n{random_tensor_B}")

# is Tensor A = Tensor B            # SHOULD BE NOT
random_tensor_A == random_tensor_B  # All False
````

````python
# Set the random seed
SEED = 42
torch.manual_seed(SEED)

random_tensor_C = torch.rand(size=(5,6))

# Have to reset the seed each time you create an new tensor
# otherwise it will be different
torch.random.manual_seed(SEED)
random_tensor_D = torch.rand(size=(5,6))

print(f"Tensor C:\n{random_tensor_C}\n")
print(f"Tensor D:\n{random_tensor_D}\n")

# Is C = D
random_tensor_C == random_tensor_D
````

### PyTorch Random Seed & PRNG State Mechanics

#### 1. إزاي الـ PRNG شغال داخلياً؟ (How PRNG Works)
* `torch.manual_seed(SEED)` بتحدد الـ **Initial State** ($S_0$) لمولد أرقام عشوائية حتمي رياضياً (Deterministic PRNG).
* كل استدعاء لأي Random function مثل `torch.rand()` بيسحب أرقام و**بيقدّم مؤشر الـ State للأمام** تلقائياً ($S_0 \to S_1 \to S_2$).
* الـ Sequence كامل حتمي (Deterministic): إعادة تشغيل الكود من $S_0$ هتنتج نفس تسلسل الأرقام بالمللي في كل Run.

---

#### 2. تثبيت الـ Seed مرة واحدة مقابل الـ Resetting
* **إعادة عمل Reset للـ Seed قبل كل Tensor:**
  * بترجع الـ Generator لنقطة البداية $S_0$ مع كل استدعاء.
  * بتخلي الـ Tensors التالية تسحب نفس الأرقام الأولى وتطلع نسخ مكررة طبق الأصل.
  * بتدمر الـ Symmetry Breaking أثناء تهيئة أوزان الـ Neural Networks وبتعطّل الـ Data Shuffling.
* **تحديد الـ Seed مرة واحدة في البداية (Standard Practice):**
  * بنحدد $S_0$ مرة واحدة فقط في أول الكود أو الـ Notebook.
  * `tensor1` بيسحب من مسار ($S_0 \to S_1$)، و`tensor2` بيكمل سحب طبيعي من ($S_1 \to S_2$).
  * الـ Tensors بتكون قيمها **مختلفة عن بعضها**، لكن التجربة كاملة بتفضل **100% Reproducible** ومستحيل تتغير قيمها بين الرنات.

---

#### 3. كود التنفيذ العملي (Implementation Pattern)

```python
import torch

# 1. بنحدد الـ seed مرة واحدة بس في الأول
SEED = 0
torch.manual_seed(SEED)

# 2. الاستدعاءات المتتالية بتولد Tensors مختلفة لكنها حتمية وثابتة كل Run
tensor_A = torch.rand(7, 7)  # بيسحب قيم الـ State: S0 -> S1
tensor_B = torch.rand(1, 7)  # بيكمل سحب قيم الـ State: S1 -> S2

# 3. عملية الضرب حتمية والناتج ثابت تماماً عند إعادة التشغيل
output = torch.mm(tensor_A, tensor_B.T)
print(output, output.shape)

## 2. Putting Tensor (and Models) on GPU

>Note: Putting a tensor on GPU using to(device) (e.g. `some_tensor.to(device)`) returns a copy of that tensor,<br>
>e.g. the same tensor will be on CPU and GPU. To overwrite tensors, reassign them:<br><br>
>`some_tensor = some_tensor.to(device)`

````python
# Creating a tensor (default on CPU)
tensor = torch.tensor([1, 2, 3])
print(f"{tensor}, {tensor.device}")

# Move the tensor to GPU
tensor_on_GPU = tensor.to(device)
tensor_on_GPU
````
